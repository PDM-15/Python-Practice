```python
def find_hcf_recursive(num1, num2):
    if num2 == 0:
        return num1
    return find_hcf_recursive(num2, num1 % num2)

# Example usage:
print(f"HCF is: {find_hcf_recursive(48, 18)}")
```

```python
def find_hcf_brute_force(a, b):
    # Choose the smaller number as the upper limit for factors
    smaller = min(a, b)

    # Check numbers backwards from 'smaller' down to 1
    for i in range(smaller, 0, -1):
        if (a % i == 0) and (b % i == 0):
            return i

# Example usage:
print(f"HCF is: {find_hcf_brute_force(48, 18)}")  # Output: 6
```

Print first 25 prime numbers -> Sieve of Eratosthenes


```python
#simplest way
def is_prime(num):
    """Checks if a single number is prime."""
    if num <= 1:
        return False

    # Check for factors from 2 up to the number - 1
    for i in range(2, num):
        if num % i == 0:
            return False  # Found a factor, so it is not prime

    return True  # No factors found, it is prime

# Define your range boundaries
start_range = 0
end_range = 25

print(f"Prime numbers between {start_range} and {end_range}:")

# Loop through the specified range (inclusive of end_range)
for current_num in range(start_range, end_range + 1):
    if is_prime(current_num):
        print(current_num, end=" ")

```

```python
#fastest way
def sieve_of_eratosthenes(limit):
    # Create a boolean array "prime[0..limit]" and initialize
    # all entries it as True.
    primes_mask = [True] * (limit + 1)
    primes_mask[0] = primes_mask[1] = False # 0 and 1 are not prime

    p = 2
    while p * p <= limit:
        # If primes_mask[p] is not changed, then it is a prime
        if primes_mask[p]:
            # Update all multiples of p starting from p*p
            for i in range(p * p, limit + 1, p):
                primes_mask[i] = False
        p += 1

    # Collect all numbers that are still marked as True
    prime_numbers = [num for num, is_prime in enumerate(primes_mask) if is_prime]
    return prime_numbers

# Example usage:
print(sieve_of_eratosthenes(25))
```

Write a program to check if a list is in ascending order or not

```python
Lorder = [1,2,3,4,5,5,6,7,8]
min,max = Lorder[0], Lorder[-1]

for i in range(0, len(Lorder)):
  if Lorder[i] < min:
    min = Lorder[i]
  if Lorder[i] > max:
    max = Lorder[i]

if (min == Lorder[0]) and (max == Lorder[-1]):
  print("In Ascending order")
elif(min == Lorder[-1]) and (max == Lorder[0]):
  print("In descending order")
else:
  print("In Random order")
```

Write a program that can sort a given unsorted list. Dont use any built in function for sorting

```python
L_sort = [23, 56, 43, 55, 99, 45, 67]

n = len(L_sort)
for i in range(n):
  for j in range(0, n - i - 1):
    if L_sort[j] > L_sort[j + 1]:
      L_sort[j], L_sort[j + 1] = L_sort[j + 1], L_sort[j]

print(L_sort)
```

Write a program to swap the key value pair for max and min values
Eg if the dict is like this {‘a’:1,’b’:2,’c’:3}

Output should be {a:3,b:2,c:1}

```python
my_dict = {"a": 1, "b": 2, "c": 3}

max_key = max(my_dict, key=my_dict.get)
min_key = min(my_dict, key=my_dict.get)

my_dict[max_key], my_dict[min_key] = my_dict[min_key], my_dict[max_key]

print(my_dict)
```

Write a program to find histogram of a given set of numbers. Take bin size from user. Print the result in the form of a dictionary.

```python
histogram = {}
num = [2,4,5,9,12,17,22,25]

bin_size = int(input("Enter the bin size:"))

for num in range(0, len(num)):
  bin_start = (num // bin_size) * bin_size
  bin_end = bin_start + bin_size
  bin_label = f"{bin_start} - {bin_end}"
  histogram[bin_label] = histogram.get(bin_label,0)+1

print(histogram)
```

Write a function that accepts a string and returns the number of upper case chars and lower case chars as a dictionary

```python
mydict = {'upper':0, 'lower':0}

def sstr(strng):
  for _ in range(0,len(strng)):
    if strng[_].isupper():
      mydict['upper'] += 1
    elif strng[_].islower():
      mydict['lower'] += 1
    else:
      pass
  return mydict

strng = "A quick lazy fox. It jumps. Over the fence"
print(sstr(strng))
```

Write a program that accepts neighbors(set of 2D co-ordinates) and a point(single 2D co-ordinate) and tells nearest neighbor(in terms of euclidean distance)

```python
set_2d = [2,5]
set_2d1 = [3,8]
point = 6

euclidean_dist = ((point -set_2d[0])**2 + (point - set_2d[1])**2) ** 0.5
euclidean_dist1 = ((point -set_2d1[0])**2 + (point - set_2d1[1])**2) ** 0.5

if euclidean_dist < euclidean_dist1:
  print("Nearest neighbor:", set_2d)
else:
  print("Nearest neighbour:", set_2d1)
```

Write a function that accepts a number and returns it’s factorial. You can not use any loop

```python
def fact(n):
  if n<0:
    return "invalid input"
  elif n == 0:
    return 1
  else:
    return n * fact(n-1)

print(fact(5))
```
